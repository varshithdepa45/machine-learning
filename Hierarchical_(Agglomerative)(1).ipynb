```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import AgglomerativeClustering
from scipy.cluster.hierarchy import dendrogram, linkage

# Dataset
X = np.array([
    [2, 10],  # A1
    [2, 5],   # A2
    [8, 4],   # A3
    [5, 8],   # B1
    [7, 5],   # B2
    [6, 4],   # B3
    [1, 2],   # C1
    [4, 9]    # C2
])

labels_names = ['A1', 'A2', 'A3', 'B1', 'B2', 'B3', 'C1', 'C2']

# Agglomerative Clustering
hc = AgglomerativeClustering(
    n_clusters=3,
    linkage='ward'
)

cluster_labels = hc.fit_predict(X)

# Display Cluster Assignments
print("Cluster Assignments:")
for i in range(len(X)):
    print(f"{labels_names[i]} {tuple(X[i])} -> Cluster {cluster_labels[i]}")

# Scatter Plot
plt.figure(figsize=(8, 6))
plt.scatter(X[:, 0], X[:, 1], c=cluster_labels, cmap='rainbow', s=100)

for i, txt in enumerate(labels_names):
    plt.annotate(txt, (X[i, 0], X[i, 1]))

plt.title("Hierarchical Agglomerative Clustering")
plt.xlabel("X Coordinate")
plt.ylabel("Y Coordinate")
plt.grid(True)
plt.show()

# Dendrogram
Z = linkage(X, method='ward')

plt.figure(figsize=(10, 5))
dendrogram(Z, labels=labels_names)
plt.title("Dendrogram")
plt.xlabel("Data Points")
plt.ylabel("Euclidean Distance")
plt.grid(True)
plt.show()
```
